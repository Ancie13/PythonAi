HOW TO MAKE A MINT: THE
CRYPTOGRAPHY OF ANNONYMOUS
ELECTRONIC CASH

```python
import numpy as np
N = 11
x = np.array(range(1, 100), dtype="object")
```

```python
x % N
```

```python
2 ** x % N
```

```python
10 ** x % N
```

```python
2 ** 21 % N
```

```python
s = 3
v = 7
M = 5
```

```python
C = M ** s % N
C
```

```python
m = C ** v % N
print(m, m == M)
```
