- Виведіть масив у читабельному вигляді (у вигляді таблиці).

- Визначте загальне середнє значення масиву.
- Виведіть усі результати в зручному вигляді (наприклад: "Суми за рядками: ...","Суми за стовпцями: ...").

```python
arr2 = np.random.randint(1, 50, size=(4,4))
```

```python
# Виведіть масив у читабельному вигляді (у вигляді таблиці).
```

```python
print(arr2)
```

```python
print(*(a for a in arr2), sep='\n')
```

```python
# Знайдіть суму елементів за кожним рядком і за кожним стовпчиком (використовуйте параметр axis).
print(np.sum(arr2, axis=0)) # axis=0 - За стовпчиками 
print(np.sum(arr2, axis=1)) # axis=1 - ЗА рядками 
```

```python
# zip(arr1,arr2) - поєднує множини за принципом [ [arr1[0], arr2[0]], [arr1[1], arr2[1]], ... ]
# map(fun, arr) - застосовує функцію fun до кожного елемента arr
# " ".join(arr) - поєднує всі елементи arr через пробіл (вимагає, щоб елементи arr були рядками)

print(
    *( f"{x} | {y}" for x, y in zip( arr2, np.sum(arr2, axis=1) ) ),
    "----------------",
    # f"{np.sum(arr2, axis=0)} {np.mean(arr2)}",
    " ".join(map(str, (np.sum(arr2, axis=0), "|", np.mean(arr2)))),
    sep='\n'

)
```

- Застосуйте операцію reshape, щоб отримати масив 5×3.

```python
arr3 = np.random.randint(1, 14, size=(3,5))
print(arr3)
```

```python
# Виконайте транспонування масиву (рядки стають стовпцями, а стовпці — рядками).
np.transpose(arr3)
```

```python
# Застосуйте операцію reshape, щоб отримати масив 5×3.
np.reshape(arr3, shape=(5,3)) # реорганізація до нової форми, без збереження логіки розподілу
```

Створіть масив розміром 10×10, заповнений випадковими числами з плаваючою крапкою від 0 до 1.

- Знайдіть мінімальне та максимальне значення в усьому масиві.
- Визначте індекси мінімального та максимального елементів (наприклад: (рядок, стовпець)).
- Виведіть сам масив і окремо знайдені значення разом з їхніми індексами.

```python
arr4 = np.random.rand(10,10)
# print(*(f"{x:.3f}" for a in arr4 for x in a))
print(*(' '.join(f"{x:.3f}" for x in a) for a in arr4), sep='\n')

# ' '.join(f"{x:.3f}" for x in arr4[0])
```

```python
# Визначте індекси мінімального та максимального елементів (наприклад: (рядок, стовпець)).
min_index = np.argmin(arr4)
print(min_index)
max_index = np.argmax(arr4)
max_index
```

```python
xy = np.unravel_index(min_index, arr4.shape)
print("мінімальний елемент знаходиться за індексами:", ",".join(map(str, xy)))
xy2 = np.unravel_index(max_index, arr4.shape)
print("максимальний елемент знаходиться за індексами:", ",".join(map(str, xy2)))
```
