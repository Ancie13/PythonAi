# Побудова мовних моделей та алгоритми Seq2Seq
## Бізнес-контекст
Компанія SupportFlow AI автоматизує відповіді служби підтримки клієнтів.
Вам необхідно побудувати прототип моделі, яка здатна генерувати короткі текстові відповіді на основі вхідного запиту користувача.
Під час завдання ви вивчите, як працюють рекурентні мережі (RNN, LSTM) і архітектури Transformer.

## Завдання 1
Підготуйте дані для навчання.

- Створіть невеликий набір пар запитання-відповідь (10 - 15 прикладів), наприклад:
- "How can I reset my password?" → "You can reset it through the account settings".
- Приведіть тексти до нижнього регістру та видаліть пунктуацію.
- Розділіть дані на токени і створіть словник (token-to-index).
- Визначте максимальну довжину вхідної та вихідної послідовності.

```python
qa_pairs = [
    ("How can I reset my password?", "You can reset it through the account settings"),
    ("Where can I find my invoice?", "Invoices are available in the billing tab"),
    ("How do I delete my account?", "Please contact our support team to close your account"),
    ("Can I change my username?", "Usernames can be updated in your profile details"),
    ("Why is my account locked?", "Your account is locked due to too many failed login attempts"),
    ("How do I enable two factor authentication?", "You can enable 2FA in the security settings menu"),
    ("Where is the user manual?", "The user manual can be downloaded from our help center"),
    ("How do I update the application?", "Go to the settings menu and click check for updates"),
    ("Can I cancel my subscription?", "You can cancel your subscription in the billing section"),
    ("How do I change my email address?", "Update your email address in the account settings info"),
    ("Why is the app crashing?", "Please clear your cache and restart the application"),
    ("How can I contact support?", "You can open a live chat or send us an email"),
    ("Where do I enter a promo code?", "Apply your coupon code during the checkout process"),
    ("How do I export my data?", "Click the export button in the data privacy tab"),
    ("Can I use this on multiple devices?", "Yes you can log in on up to three devices simultaneously")
]
```

```python
# Приведіть тексти до нижнього регістру та видаліть пунктуацію.
# Розділіть дані на токени і створіть словник (token-to-index).
import re
def prep(s:str):
    return re.sub(r'[^\w\s]+', '', s.lower())

SOS_token = 0
EOS_token = 1

from nltk.tokenize import word_tokenize
questions = []
answers = []
token_to_index = {
    "SOS": SOS_token,
    "EOS": EOS_token
}
index_to_token = {
    SOS_token: "SOS",
    EOS_token: "EOS"
}

max_len = 0
for q, a in qa_pairs:
    q, a = prep(q), prep(a)
    questions.append(q)
    answers.append(a)
    tokens_q = word_tokenize(q)
    tokens_a = word_tokenize(a)
    
    
    for token in word_tokenize(q) + word_tokenize(a):
        if len(tokens_q) > max_len:
            max_len = len(tokens_q)
        if len(tokens_a) > max_len:
            max_len = len(tokens_a)
        if not token in token_to_index.keys():
            key = len(token_to_index)
            token_to_index[token] = key
            index_to_token[key] = token

print(token_to_index)
print('-----------')
print(index_to_token)
```

```python
MAX_LENGTH = max_len + 2

import torch

def sentence_to_tensor(s:str):
    seq = [ SOS_token, 
            *(token_to_index.get(token, SOS_token) for token in word_tokenize(prep(s))) 
    ]
    if len(seq) >= MAX_LENGTH:
        seq = seq[:MAX_LENGTH - 1] + [EOS_token]
    else:
        seq = seq + [EOS_token for _ in range(MAX_LENGTH - len(seq))]
    return torch.tensor([seq])


print(sentence_to_tensor(questions[0]))
```

## Завдання 2
Реалізуйте просту RNN-модель для генерації відповідей.

- Використовуйте бібліотеку torch.nn.
- Створіть модель на основі nn.RNN з параметрами:
     - input_size = розмір словника;
     - hidden_size = 128;
     - num_layers = 1.
- Використовуйте шар nn.Linear для отримання вихідної послідовності.
     - Функція втрат — CrossEntropyLoss, оптимізатор — Adam (lr = 0.001).
     - Навчіть модель 10 епох.

```python
import torch.nn.functional as F
import torch.nn as nn


class EncoderRNN(nn.Module):
    def __init__(self, input_size, hidden_size=128, dropout_p=0.1):
        super(EncoderRNN, self).__init__()
        self.hidden_size = hidden_size

        self.embedding = nn.Embedding(input_size, hidden_size)
        self.rnn = nn.RNN(hidden_size, hidden_size, num_layers=1, batch_first=True)
        self.dropout = nn.Dropout(dropout_p)

    def forward(self, input):
        embedded = self.dropout(self.embedding(input))
        output, hidden = self.rnn(embedded)
        return output, hidden

class DecoderRNN(nn.Module):
    def __init__(self, output_size, hidden_size=128):
        super(DecoderRNN, self).__init__()
        self.embedding = nn.Embedding(output_size, hidden_size)
        self.rnn = nn.RNN(hidden_size, hidden_size, batch_first=True)
        self.out = nn.Linear(hidden_size, output_size)

    def forward(self, encoder_outputs, encoder_hidden, num_layers=1, target_tensor=None):
        batch_size = encoder_outputs.size(0)
        decoder_input = torch.empty(batch_size, 1, dtype=torch.long).fill_(SOS_token)
        decoder_hidden = encoder_hidden
        decoder_outputs = []

        for i in range(MAX_LENGTH):
            decoder_output, decoder_hidden  = self.forward_step(decoder_input, decoder_hidden)
            decoder_outputs.append(decoder_output)

            if target_tensor is not None:
                # Teacher forcing: Feed the target as the next input
                decoder_input = target_tensor[:, i].unsqueeze(1) # Teacher forcing
            else:
                # Without teacher forcing: use its own predictions as the next input
                _, topi = decoder_output.topk(1)
                decoder_input = topi.squeeze(-1).detach()  # detach from history as input

        decoder_outputs = torch.cat(decoder_outputs, dim=1)
        decoder_outputs = F.log_softmax(decoder_outputs, dim=-1)
        return decoder_outputs, decoder_hidden, None # We return `None` for consistency in the training loop

    def forward_step(self, input, hidden):
        output = self.embedding(input)
        output = F.relu(output)
        output, hidden = self.rnn(output, hidden)
        output = self.out(output)
        return output, hidden

class RnnModel:
    def __init__(self, token_to_index, index_to_token):
        lr = 0.001
        input_size = len(token_to_index)
        self.encoder = EncoderRNN(input_size)
        self.decoder = DecoderRNN(output_size=input_size)
        self.enc_optimizer = torch.optim.Adam(self.encoder.parameters(), lr=lr)
        self.dec_optimizer = torch.optim.Adam(self.decoder.parameters(), lr=lr)
        self.loss_f = torch.nn.CrossEntropyLoss()

    def fit(self, q, a, epochs=10):
        for epoch in range(epochs):
            total_loss = 0
            for i in range(len(q)):
                input_tensor = q[i]
                target_tensor = a[i]
                self.enc_optimizer.zero_grad()
                self.dec_optimizer.zero_grad()
                enc_out, enc_h = self.encoder(input_tensor)
                dec_out, _, _ = self.decoder(enc_out, enc_h, target_tensor)
                loss = self.loss_f(
                    dec_out.view(-1, dec_out.size(-1)),
                    target_tensor.view(-1)
                )
                loss.backward()
                self.enc_optimizer.step()
                self.dec_optimizer.step()
                total_loss += loss.item()
            print(epoch + 1, total_loss)
        return self
```

```python
question_tensors = [sentence_to_tensor(q) for a in questions]
answer_tensors = [sentence_to_tensor(a) for a in questions]
```

```python
model = RnnModel(token_to_index, index_to_token)
model.fit(question_tensors, answer_tensors)
```
