```python
# Очистіть текст від пунктуації та приведіть до нижнього регістру.
import re
feedbacks_cleared = {
    "positive": [re.sub(r"[^\w\s]", "", f.lower()) for f in feedbacks["positive"]],
    "negative": [re.sub(r"[^\w\s]", "", f.lower()) for f in feedbacks["negative"]],
}

feedbacks_cleared
```

## Завдання 3
Видаліть стоп-слова і проаналізуйте частотність.

- Завантажте список англійських стоп-слів nltk.corpus.stopwords.words('english').
- Видаліть із токенів усі стоп-слова.
- Підрахуйте частоту слів, що залишилися, за допомогою collections.Counter.
- Виведіть 10 найчастіших слів та їхню кількість.

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
print(len(stop_words))
print(stop_words[:10])
```

```python
# Видаліть із токенів усі стоп-слова.
feedbacks_stop = {
    "positive": [ token for token in feedback_tokens["positive"] if not token in stop_words ],
    "negative": [ token for token in feedback_tokens["negative"] if not token in stop_words ]
}
print(len(feedback_tokens["positive"]), len(feedback_tokens["negative"]) )
print(len(feedbacks_stop["positive"]), len(feedbacks_stop["negative"]) )
```

```python
from collections import Counter
feedbacks_freqs = {
    "positive": Counter(feedbacks_stop["positive"]),
    "negative": Counter(feedbacks_stop["negative"])
}
md( "\n".join((
    "|Слово(+)|Кількість|Слово(-)|Кількість",
    "-|-|-|-",
    *(f"{pos[0]}|{pos[1]}|{neg[0]}|{neg[1]}"
      for pos, neg in zip(
          sorted(
              list(feedbacks_freqs["positive"].items()),
              key=lambda x: x[1],
              reverse=True
          )[:10],
          sorted(
              list(feedbacks_freqs["negative"].items()),
              key=lambda x: x[1],
              reverse=True
          )[:10]
      )
     )
)) )
```
