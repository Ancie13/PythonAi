- Сформуйте таблицю мінімум на 30 днів із колонками "date", "users", "sessions", "revenue".
- Розрахуйте кореляційну матрицю для цих метрик.
- Побудуйте діаграми розсіювання для пар: users-sessions, users-revenue, sessions-revenue.
- Побудуйте лінійний графік "revenue" за датами.
- Виведіть матрицю та всі графіки.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Сформуйте таблицю мінімум на 30 днів із колонками "date", "users", "sessions", "revenue".
dates = pd.date_range(start="2026-07-01", periods=30)

data = {
    "date": dates,
    "users": np.random.randint(500, 2000, size=30),
    "sessions": np.random.randint(800, 3500, size=30),
    "revenue": np.random.randint(5000, 25000, size=30)
}

df = pd.DataFrame(data)

print(df)
```

```python
# Розрахуйте кореляційну матрицю для цих метрик.
matrix = df[["users", "sessions", "revenue"]].corr()
print(matrix)
```

```python
# Побудуйте діаграми розсіювання для пар: users-sessions, users-revenue, sessions-revenue.
_, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(8, 6))

ax1.scatter(df[["users"]],  df[["sessions"]])
ax1.set_title("users-sessions")
ax1.set_ylabel("sessions")
ax1.set_xlabel("users")
ax1.tick_params(axis="x", rotation=45)
ax1.grid(True)

ax2.scatter(df[["users"]],  df[["revenue"]])
ax2.set_title("users-revenue")
ax2.set_ylabel("revenue")
ax2.set_xlabel("users")
ax2.tick_params(axis="x", rotation=45)
ax2.grid(True)

ax3.scatter(df[["sessions"]],  df[["revenue"]])
ax3.set_title("sessions-revenue")
ax3.set_ylabel("revenue")
ax3.set_xlabel("sessions")
ax3.tick_params(axis="x", rotation=45)
ax3.grid(True)

ax4.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Побудуйте лінійний графік "revenue" за датами.
_, ax = plt.subplots(figsize=(8, 6))

ax.plot(dates, df[["revenue"]])
ax.set_title("Conversion")
ax.set_ylabel("Conversion")
ax.tick_params(axis="x", rotation=45)
ax.grid(True)

plt.show()
```
