# Завдання 1
Є дані про планові та фактичні продажі за місяцями. Потрібно порівняти два ряди на одному графіку.

- Створіть список місяців і два списки значень: "План" і "Факт".
- Побудуйте один графік із двома лініями.
- Додайте підписи осей і заголовок.
- Додайте легенду для ліній.
- Відобразіть графік на екрані.

```python
import matplotlib.pyplot as plt
```

```python
# Створіть список місяців і два списки значень: "План" і "Факт".
months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

plan = [
    10000,
    12000,
    15000,
    14000,
    16000,
    18000,
    17000,
    19000,
    21000,
    20000,
    23000,
    25000
]

actual = [
    9500,
    12500,
    14200,
    15000,
    15800,
    17500,
    18200,
    18500,
    22000,
    19500,
    24000,
    26500
]
```

```python
# Побудуйте один графік із двома лініями.
# Додайте підписи осей і заголовок.
# Додайте легенду для ліній.
_, ax = plt.subplots()
ax.plot(months, plan, label="Plan")
ax.plot(months, actual, label="Actual")
ax.set_title("Annual Sales Plan")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.tick_params(axis="x", rotation=45)
ax.legend()
ax.grid(True)
plt.show()
```

## Завдання 2
Дано значення вікових груп 100 осіб. Потрібно показати розподіл.

- Створіть список (або масив) зі 100 цілих значень віку.
- Побудуйте гістограму розподілу.
- Додайте вертикальну лінію середнього значення.
- Підпишіть осі та додайте заголовок.
- Відобразіть графік на екрані.

```python
# Створіть список (або масив) зі 100 цілих значень віку.
import numpy as np
ages = np.random.randint(1, 101, size=100)
print(ages)
```

```python
# Побудуйте гістограму розподілу.
# Додайте вертикальну лінію середнього значення.
# Підпишіть осі та додайте заголовок.
# Відобразіть графік на екрані.
mean_age = np.mean(ages)
_, ax = plt.subplots()
ax.hist(ages, bins=10)
ax.axvline(mean_age, linestyle="--", color="red",
           label=f"Mean: {mean_age:.1f}")
ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Number of People")
plt.show()
```

```python
dates = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

temperature = [22, 24, 21, 25, 27, 26, 23]

humidity = [65, 60, 70, 55, 50, 52, 68]
```

```python
# Побудуйте лінійний графік температури.
# На тій самій області побудуйте графік вологості.
# Додайте легенду, підписи осей і заголовок.
# Поверніть підписи дат на осі X для читабельності та відобразіть графік.
_, ax = plt.subplots(figsize=(5, 3))
ax.plot(dates, humidity, label="Humidity")
ax.plot(dates,temperature, label="Temperature")
ax.set_title("Temperature and Humidity During the Week")
ax.set_xlabel("Day")
ax.set_ylabel("Value")
ax.tick_params(axis="x", rotation=45)
ax.legend()
ax.grid(True)
plt.show()
```

## Завдання 5
Є погодинні дані навантаження сервера за добу. Потрібно показати зміни і виділити область між кривою і віссю X.

- Створіть список годин (0-23) і відповідні значення навантаження.
- Побудуйте лінійний графік навантаження.
- Зафарбуйте область під кривою.
- Додайте сітку, підписи осей і заголовок.
- Відобразіть графік на екрані.

```python
# Створіть список годин (0-23) і відповідні значення навантаження.
hours = list(range(24))

load = [
    25, 22, 20, 18, 17, 19,
    24, 32, 45, 52, 58, 61,
    65, 68, 64, 60, 57, 63,
    70, 75, 68, 55, 42, 32
]
```

```python
# Побудуйте лінійний графік навантаження.
# Зафарбуйте область під кривою.
# Додайте сітку, підписи осей і заголовок.
# Відобразіть графік на екрані.
_, ax = plt.subplots(figsize=(5, 3))
ax.plot(hours, load, label="load")
ax.set_title("Server load during the day")
ax.set_xlabel("Hours")
ax.set_ylabel("Value")
ax.fill_between(hours, load)
ax.grid(True)
plt.show()
```

## Завдання 6
Дано чотири набори метрик продукту: конверсія, утримання, середній чек, кількість замовлень. Потрібно показати їх на окремих панелях.

- Створіть чотири набори числових значень за однією шкалою часу.
- Створіть сітку з чотирьох підграфіків 2×2.
- На кожному підграфіку побудуйте відповідний графік.
- Додайте заголовки для кожного підграфіка та загальний заголовок.
- Відобразіть результат на екрані.

```python
# Створіть чотири набори числових значень за однією шкалою часу.
days = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

conversion = [3.2, 3.5, 3.1, 4.0, 4.3, 4.7, 4.1]

retention = [72, 74, 71, 76, 78, 80, 77]

average_order = [25, 27, 24, 30, 32, 35, 31]

orders = [120, 135, 128, 150, 165, 180, 155]
```

```python
# Створіть сітку з чотирьох підграфіків 2×2.
# На кожному підграфіку побудуйте відповідний графік.
# Додайте заголовки для кожного підграфіка та загальний заголовок.
# Відобразіть результат на екрані.
_, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(8, 6))

ax1.plot(days, conversion, label="Conversion")
ax1.set_title("Conversion")
ax1.set_ylabel("Conversion")
ax1.tick_params(axis="x", rotation=45)
ax1.grid(True)

ax2.plot(days, retention, label="Retention")
ax2.set_title("Retention")
ax2.set_ylabel("Retention")
ax2.tick_params(axis="x", rotation=45)
ax2.grid(True)

ax3.plot(days, average_order, label="Average Order")
ax3.set_title("Average Order")
ax3.set_ylabel("Average Order")
ax3.tick_params(axis="x", rotation=45)
ax3.grid(True)

ax4.plot(days, orders, label="Orders")
ax4.set_title("Orders")
ax4.set_xlabel("Days")
ax4.set_ylabel("Orders")
ax4.tick_params(axis="x", rotation=45)
ax4.grid(True)

plt.tight_layout()
plt.show()
```
