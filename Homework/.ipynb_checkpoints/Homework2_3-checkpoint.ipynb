# Завдання 1
Є дані про планові та фактичні продажі за місяцями. Потрібно порівняти два ряди на одному графіку.

- Створіть список місяців і два списки значень: "План" і "Факт".
- Побудуйте один графік із двома лініями.
- Додайте підписи осей і заголовок.
- Додайте легенду для ліній.
- Відобразіть графік на екрані.

```python
import matplotlib.pyplot as plt
```

```python
# Створіть список місяців і два списки значень: "План" і "Факт".
months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

plan = [
    10000,
    12000,
    15000,
    14000,
    16000,
    18000,
    17000,
    19000,
    21000,
    20000,
    23000,
    25000
]

actual = [
    9500,
    12500,
    14200,
    15000,
    15800,
    17500,
    18200,
    18500,
    22000,
    19500,
    24000,
    26500
]
```

```python
# Побудуйте один графік із двома лініями.
# Додайте підписи осей і заголовок.
# Додайте легенду для ліній.
_, ax = plt.subplots()
ax.plot(months, plan, label="Plan")
ax.plot(months, actual, label="Actual")
ax.set_title("Annual Sales Plan")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.tick_params(axis="x", rotation=45)
ax.legend()
ax.grid(True)
plt.show()
```

## Завдання 2
Дано значення вікових груп 100 осіб. Потрібно показати розподіл.

- Створіть список (або масив) зі 100 цілих значень віку.
- Побудуйте гістограму розподілу.
- Додайте вертикальну лінію середнього значення.
- Підпишіть осі та додайте заголовок.
- Відобразіть графік на екрані.

```python
# Створіть список (або масив) зі 100 цілих значень віку.
import numpy as np
ages = np.random.randint(1, 101, size=100)
print(ages)
```

```python
# Побудуйте гістограму розподілу.
# Додайте вертикальну лінію середнього значення.
# Підпишіть осі та додайте заголовок.
# Відобразіть графік на екрані.
mean_age = np.mean(ages)
_, ax = plt.subplots()
ax.hist(ages, bins=10)
ax.axvline(mean_age, linestyle="--", color="red",
           label=f"Mean: {mean_age:.1f}")
ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Number of People")
plt.show()
```

## Завдання 3
Є результати іспитів за трьома групами студентів. Потрібно порівняти розкид оцінок між групами.

- Створіть три набори числових результатів (по одній вибірці на групу).
- Побудуйте графік boxplot для трьох груп поруч.
- Підпишіть групи по осі X.
- Додайте підписи осей і заголовок.
- Відобразіть графік на екрані.

```python
group1 = [78, 82, 85, 90, 76, 88, 91, 84, 79, 87]
group2 = [60, 75, 95, 82, 68, 90, 55, 88, 72, 98]
group3 = [80, 81, 79, 82, 80, 83, 81, 79, 84, 80]
```

```python
# Побудуйте графік boxplot для трьох груп поруч.
# Підпишіть групи по осі X.
# Додайте підписи осей і заголовок.
# Відобразіть графік на екрані.
_, ax = plt.subplots()
ax.boxplot([group1, group2, group3])
ax.set_xticklabels(["group1", "group2", "group3"])
ax.set_title("Exam Scores Distribution")
ax.set_xlabel("Group")
ax.set_ylabel("Score")
plt.show()
```
