## Завдання 1

- Створіть масив із 20 цілих чисел у діапазоні від 0 до 50.
- Виведіть вихідний масив.
- Запитайте у користувача число (порогове значення).
- Визначте кількість елементів масиву, що перевищують введене користувачем значення, і виведіть результат.
- Знайдіть максимальне значення в масиві та позицію його першої появи, виведіть обидва результати.
- Відсортуйте масив за спаданням і виведіть відсортований масив.

```python
import numpy as np
```

```python
# Створіть масив із 20 цілих чисел у діапазоні від 0 до 50.
# Виведіть вихідний масив.
arr1 = np.random.randint(0, 51, size=20)
print(arr1)
```

```python
# Запитайте у користувача число (порогове значення).
usersNum = int(input("Enter threshold value:"))
```

```python
# Визначте кількість елементів масиву, що перевищують введене користувачем значення, і виведіть результат.
sumHigherThanUsersNum = sum(1 for num in arr1 if num > usersNum)
print(sumHigherThanUsersNum)
```

```python
# Знайдіть максимальне значення в масиві та позицію його першої появи, виведіть обидва результати.
MaxValueOfArr1 = max(arr1)
print("Max value:", MaxValueOfArr1)
print("First appearance index: ", np.where(arr1 == MaxValueOfArr1)[0])
```

```python
# Відсортуйте масив за спаданням і виведіть відсортований масив.
print(np.sort(arr1)[::-1])
```

## Завдання 2

- Запитайте в користувача верхню та нижню межу діапазону чисел.
- Створіть квадратну матрицю 5×5 з цілими значеннями в цьому діапазоні.
- Виведіть вихідну матрицю.
- Отримайте головну діагональ і виведіть діагональні елементи.
- Підрахуйте суму елементів головної діагоналі та виведіть результат.
- Обнуліть елементи строго вище головної діагоналі та виведіть змінену матрицю.

```python
# Запитайте в користувача верхню та нижню межу діапазону чисел.
topLimit = int(input("Enter top Limit:"))
bottomLimit = int(input("Enter bottom Limit:"))
```

```python
# Створіть квадратну матрицю 5×5 з цілими значеннями в цьому діапазоні.
# Виведіть вихідну матрицю.
arr2 = np.random.randint(bottomLimit+1, topLimit, size=(5,5))
print(arr2)
```

```python
# Отримайте головну діагональ і виведіть діагональні елементи.
primaryDiag = np.diag(arr2)
primaryDiag
```

```python
# Підрахуйте суму елементів головної діагоналі та виведіть результат.
sum(primaryDiag)
```

```python
# Обнуліть елементи строго вище головної діагоналі та виведіть змінену матрицю.
np.tril(arr2)
```

- Перетворіть послідовність на матрицю розміром 6×5 і виведіть її.

```python
# Запитайте у користувача початок і кінець діапазону чисел.
topLimit = int(input("Enter top Limit:"))
bottomLimit = int(input("Enter bottom Limit:"))
```

## Завдання 4

- Запитайте в користувача діапазон значень (нижню і верхню межу), включно з від'ємними числами.
- Створіть одновимірний масив із 15 випадкових цілих чисел у вказаному діапазоні.
- Виведіть вихідний масив.
- Виділіть усі від'ємні елементи та виведіть їх як окремий масив.
- Створіть змінену копію масиву, в якій від'ємні елементи замінено на нулі, і виведіть її.
- Підрахуйте кількість нульових елементів у зміненому масиві та виведіть результат.

```python
# Запитайте в користувача діапазон значень (нижню і верхню межу), включно з від'ємними числами.
topLimit = int(input("Enter top Limit (including negative numbers):"))
bottomLimit = int(input("Enter bottom Limit (including negative numbers):"))
```

```python
# Створіть одновимірний масив із 15 випадкових цілих чисел у вказаному діапазоні.
# Виведіть вихідний масив.
arr4 = np.random.randint(bottomLimit, topLimit+1, size=15)
arr4
```

```python
# Виділіть усі від'ємні елементи та виведіть їх як окремий масив.
negativeArr4 = []
for num in arr4:
    if num < 0:
        negativeArr4.append(num)
negativeArr4
```

```python
# Створіть змінену копію масиву, в якій від'ємні елементи замінено на нулі, і виведіть її.
arr4WithoutNegatives = np.where(arr4 < 0, 0, arr4)
arr4WithoutNegatives
```

```python
# Підрахуйте кількість нульових елементів у зміненому масиві та виведіть результат.
arr4SumOfZeroEllements = np.sum(arr4WithoutNegatives == 0)
arr4SumOfZeroEllements
```

## Завдання 5

- Запитайте в користувача довжину масивів.
- Створіть два одновимірні масиви вказаної довжини:
    - перший заповніть випадковими числами від 0 до 10;
    - другий заповніть випадковими числами від 10 до 20.
- Виведіть обидва вихідні масиви.
- Об'єднайте масиви в один і виведіть результат.
- Виконайте поелементне додавання та поелементну різницю вихідних масивів і виведіть обидва результати.

```python
# Запитайте в користувача довжину масивів.
arr5len = int(input("Enter array5 length:"))
arr6len = int(input("Enter array6 length:"))
```

```python
# Створіть два одновимірні масиви вказаної довжини:
#     перший заповніть випадковими числами від 0 до 10;
#     другий заповніть випадковими числами від 10 до 20.
arr5 = np.random.randint(0, 11, size=arr5len)
arr6 = np.random.randint(10, 21, size=arr6len)
print(arr5)
print(arr6)
```

```python
# Об'єднайте масиви в один і виведіть результат.
arr5 = np.append(arr5, arr6)
arr5
```

```python
# Виконайте поелементне додавання та поелементну різницю вихідних масивів і виведіть обидва результати.
n = min(len(arr5), len(arr6))

arrsSum = arr5.copy()
arrsDifference = arr5.copy()

arrsSum[:n] = arr5[:n] + arr6[:n]
arrsDifference[:n] = arr5[:n] - arr6[:n]
print("arr5 and arr6 sum", arrsSum)
print("arr5 and arr6 difference", arrsDifference)
```

## Завдання 6

- Запитайте у користувача розмірність вихідної матриці (наприклад, 4×6).
- Створіть матрицю з послідовно зростаючих цілих чисел заданої розмірності.
- Виведіть вихідну матрицю.
- Запитайте в користувача нові розміри (наприклад, 2×12) і перетворіть матрицю в зазначену форму.
- Виведіть отриману матрицю.
- Визначте мінімальні та максимальні значення в кожному рядку нової матриці та виведіть обидва одновимірні масиви.
- Підрахуйте загальну суму всіх елементів нової матриці та виведіть результат.

```python
# Запитайте у користувача розмірність вихідної матриці (наприклад, 4×6).
matrixCols = int(input("Enter matrix Cols count:"))
matrixRows = int(input("Enter matrix Rows count:"))
```

```python
# Створіть матрицю з послідовно зростаючих цілих чисел заданої розмірності.
# Виведіть вихідну матрицю.
arr7 = np.arange(1, matrixRows * matrixCols + 1).reshape(matrixRows, matrixCols)
arr7
```

```python
# Запитайте в користувача нові розміри (наприклад, 2×12) і перетворіть матрицю в зазначену форму.
# Виведіть отриману матрицю.
matrixCols = int(input("Enter matrix Cols count:"))
matrixRows = int(input("Enter matrix Rows count:"))
arr7 = arr7.reshape(matrixRows, matrixCols)
arr7
```

```python
# Визначте мінімальні та максимальні значення в кожному рядку нової матриці та виведіть обидва одновимірні масиви.
print(np.max(arr7, axis=1))
print(np.min(arr7, axis=1))
```

```python
# Підрахуйте загальну суму всіх елементів нової матриці та виведіть результат.
print(np.sum(arr7))
```
